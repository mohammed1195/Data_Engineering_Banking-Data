```python
raw_df = spark.read \
.format("csv") \
.option("inferSchema","true") \
.option("header","true") \
.load("/user/itv026162/lendingclub/accepted_2007_to_2018Q4.csv")
```

if we observe the data, we dont have the member id which should b unique to identify
Assumptions : A user can have multiple loans, that means chances to have the same member id
------
so am going to use a hash fucntion which can consider 9 coln's with which the member id is being generated ( again this is an assumption - which is making sense as of now )

columns
1- emp_title
2- emp_length
3- home_ownership
4- annual_inc
5- zip_code
6- addr_state
7- grade
8- sub_grade
9- verification_status


```python
# working on the hash, with the 9 columns to generate a new column 
from pyspark.sql.functions import sha2, concat_ws
```

```python
new_df = raw_df.withColumn("name_sha2",sha2(concat_ws("||",*["emp_title","emp_length","home_ownership","annual_inc","zip_code","addr_state","grade","sub_grade","verification_status"]),256))
```

```python
new_df.createOrReplaceTempView("new_table") #check the last column for unique value
```

```python
spark.sql("select count(*) from new_table")
```

```python
spark.sql("select count(distinct(name_sha2)) from new_table")
```

```python
spark.sql("""select name_sha2, count(*) as total_cnt
from new_table
group by name_sha2
having total_cnt > 1
order by total_cnt desc
""")
#found an outlier where the total number of loans that are being approve to this customer is 33 times which is a kind of case to look up into
```

```python
spark.sql("""
select *
from new_table
where name_sha2 like "e3b0c44298fc1c149%"
""")
# as obsereved the all the columns are null hence the data can be easliy deleted, which we can avoid from analysis completely.
```

```python
spark.sql("""
select name_sha2 as member_id,
emp_title, emp_length, home_ownership,annual_inc, addr_state, zip_code, 'USA' as country,
grade, sub_grade, verification_status,tot_hi_cred_lim, application_type, annual_inc_joint,
verification_status_joint
from new_table""").repartition(1).write \
.option("header", True) \
.format("csv") \
.mode("overwrite") \
.option("path","/user/itv026162/lendingclub/project/raw/customers_data_csv") \
.save()
#created a subset of data from the raw data with respective coolumns as in for customer specific data, only customer information is present
```
